```python
#Fancy Indexing

arr = np.array([10, 20, 30, 40, 50])
print(arr[[0,2,4]])
```

```python
arr = np.array([100, 200, 300, 400, 500])
indices = [3, 1, 4, 0]
print(arr[indices])
```

```python
arr = np.array([10, 20, 30, 40, 50])
mask = arr > 25
print(mask)
```

```python
print(arr[mask])
```

```python
#Ufuncs

arr = np.array([1, 2, 3, 4, 5])

print(np.add(arr, 5))
print(np.square(arr))
print(np.sqrt(arr))
print(np.exp(arr))
```

```python
a = np.array([10, 20, 30])
b = np.array([1, 2, 3,4])

print(np.add(a, b))
print(np.subtract(a, b))
print(np.multiply(a, b))
print(np.divide(a, b))
```

```python
#Ufuncs with Broadcasting

arr = np.array([1, 2, 3])
print(np.add(arr,10))
```
