```python
import numpy as np
import pandas as pd

rng = np.random.default_rng(42)
N = 6000

categories = ["Apparel", "Electronics", "Home", "Footwear", "Beauty"]
cat_probs = [0.32, 0.22, 0.18, 0.18, 0.10]
payment_methods = ["COD", "Prepaid_Card", "Prepaid_UPI", "Wallet"]
pay_probs = [0.42, 0.24, 0.24, 0.10]

product_category = rng.choice(categories, size=N, p=cat_probs)
payment_method = rng.choice(payment_methods, size=N, p=pay_probs)

base_price = {
    "Apparel": (400, 2200), "Electronics": (1200, 45000), "Home": (300, 8000),
    "Footwear": (500, 4500), "Beauty": (150, 2500),
}
price_inr = np.round(np.array([rng.uniform(*base_price[c]) for c in product_category]), 0)

discount_pct = np.clip(rng.normal(22, 15, N), 0, 75)
customer_tenure_days = np.clip(rng.exponential(380, N), 1, 2500).round(0)
num_previous_orders = np.clip((customer_tenure_days / 45) + rng.normal(0, 2, N), 0, None).round(0)
base_return_rate = np.clip(rng.beta(1.5, 9, N), 0, 1)
num_previous_returns = np.round(base_return_rate * num_previous_orders).clip(0, num_previous_orders)

delivery_distance_km = np.clip(rng.gamma(3, 90, N), 2, 2200).round(1)
delivery_days = np.clip(rng.normal(4.5, 2.2, N), 1, 21).round(0)
is_weekend_order = rng.integers(0, 2, N)

rating_given = rng.integers(1, 6, N).astype(float)
missing_mask = rng.random(N) < np.where(payment_method == "COD", 0.22, 0.06)
rating_given[missing_mask] = np.nan

fit_risk_cat = np.isin(product_category, ["Apparel", "Footwear"]).astype(float)
prev_return_ratio = np.where(num_previous_orders > 0,
                              num_previous_returns / np.maximum(num_previous_orders, 1), 0)

z = (-2.2 + 1.9 * prev_return_ratio + 0.55 * fit_risk_cat
     + 0.014 * (discount_pct - 20) / 10 + 0.9 * (payment_method == "COD").astype(float)
     + 0.10 * (delivery_days - 4.5) / 2 + 0.30 * (price_inr / base_price["Electronics"][1])
     + 0.05 * is_weekend_order - 0.15 * np.tanh(customer_tenure_days / 500))
prob_return = 1 / (1 + np.exp(-z))
returned = (rng.random(N) < prob_return).astype(int)

df = pd.DataFrame({
    "order_id": np.arange(1, N + 1), "product_category": product_category,
    "price_inr": price_inr, "discount_pct": np.round(discount_pct, 1),
    "payment_method": payment_method, "customer_tenure_days": customer_tenure_days.astype(int),
    "num_previous_orders": num_previous_orders.astype(int),
    "num_previous_returns": num_previous_returns.astype(int),
    "delivery_distance_km": delivery_distance_km, "delivery_days": delivery_days.astype(int),
    "is_weekend_order": is_weekend_order, "rating_given": rating_given, "returned": returned,
})
df.to_csv("orders_dataset.csv", index=False)
print("Rows:", len(df), "| Return rate:", round(df["returned"].mean(), 4))
```

```python
#data set spliting ---test Train Split 80% traing & 20% testing

X = df_encoded.drop(columns=['returned'])
y = df_encoded['returned']

print(f'Feature matrix X:  {X.shape[0]:,} rows × {X.shape[1]} columns')
print(f'Target vector y:   {y.shape[0]:,} values')
print(f'Positive rate:     {y.mean():.1%} (1=returned)')

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size    = 0.20,
    random_state = SEED,
    stratify     = y       # ← preserves class ratio in both splits
)
print(f'\nSplit summary:')
print(f'  Train: {len(X_train):,} rows | positive rate = {y_train.mean():.1%}')
print(f'  Test:  {len(X_test):,}  rows | positive rate = {y_test.mean():.1%}')
print(f'  ✅ Class ratios are consistent across splits.')
```

```python
# feature catogerationing for the normalysation
num_features = X_train.select_dtypes(include='number').columns.tolist()
cat_features = X_train.select_dtypes(include='string').columns.tolist()
print(f'\nNumerical features ({len(num_features)}): {num_features[:4]}...')
print(f'Categorical features ({len(cat_features)}): {cat_features}')
```

```python
#Data normalization to ensures that features (columns/ (rows) with large numeric ranges do not disproportionately influence the model

num_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='median')),
    ('scaler',  StandardScaler()),
])

cat_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='most_frequent')),
    ('encoder', OneHotEncoder(handle_unknown='ignore', sparse_output=False)),
])

preprocessor = ColumnTransformer(transformers=[
    ('num', num_transformer, num_features),
    ('cat', cat_transformer, cat_features),
])

X_train_transformed = preprocessor.fit_transform(X_train)
X_test_transformed  = preprocessor.transform(X_test)

n_num_out = len(num_features)
n_cat_out = X_train_transformed.shape[1] - n_num_out

print(f'Preprocessing output shape:')
print(f'  Train: {X_train_transformed.shape}  (was {X_train.shape})')
print(f'  Test:  {X_test_transformed.shape}   (was {X_test.shape})')
print(f'  Numerical features:          {n_num_out}')
print(f'  One-hot encoded categoricals: {n_cat_out}')
print(f'  Total features after encoding: {X_train_transformed.shape[1]}')
print()
print('✅ Preprocessor fitted on training data only.')
print('   Test data transformed using training statistics — no leakage.')
```

```python
# ── Evaluation matrics ─────────────────────────────────────────────────────
def evaluate_classifier(y_true, y_pred, y_proba=None, model_name='Model'):
    acc    = accuracy_score(y_true, y_pred)
    f1_w   = f1_score(y_true, y_pred, average='weighted', zero_division=0)
    f1_b   = f1_score(y_true, y_pred, average='binary',   zero_division=0)
    auc    = roc_auc_score(y_true, y_proba) if y_proba is not None else None

    print(f'\n{"─"*52}')
    print(f'  {model_name}')
    print(f'{"─"*52}')
    print(f'  Accuracy:          {acc:.4f}  ({acc*100:.2f}%)')
    print(f'  F1 (weighted):     {f1_w:.4f}')
    print(f'  F1 (binary/pos):   {f1_b:.4f}')
    if auc: print(f'  ROC-AUC:           {auc:.4f}')
    print(f'\n  Classification Report:')
    report = classification_report(y_true, y_pred,
                                   target_names=['Stay (0)','returned(1)'],
                                   zero_division=0)
    for line in report.split('\n'):
        print(f'    {line}')
    return {'accuracy':acc,'f1_weighted':f1_w,'f1_binary':f1_b,'roc_auc':auc}

# ── Build and evaluate baseline ───────────────────────────────────────────
baseline_pipe = Pipeline([
    ('preprocessor', preprocessor),
    ('model',        DummyClassifier(strategy='most_frequent', random_state=SEED)),
])
baseline_pipe.fit(X_train, y_train)
y_pred_bl = baseline_pipe.predict(X_test)

print('What the baseline predicts for every returned:')
print(pd.Series(y_pred_bl).value_counts())
print('(Always 0 = No returned, regardless of any features)')

```

```python
#baseline model for the comparision between normal model & trained suitable model
lr_pipe = Pipeline([
    ('preprocessor', preprocessor),
    ('model', LogisticRegression(
        class_weight = 'balanced',  # corrects for 20%/80% imbalance
        max_iter     = 1000,        # allow convergence for this dataset
        random_state = 42,
        C            = 1.0,         # regularisation: smaller C = stronger
        solver       = 'lbfgs',
    ))
])
baseline_metrics = evaluate_classifier(y_test, y_pred_bl,
                                        model_name='BASELINE — DummyClassifier ()')
```

```python
#Actual ML model-RandomForestClassifire

rf_pipe = Pipeline([
    ('preprocessor', preprocessor),
    ('model', RandomForestClassifier(
        n_estimators = 100,
        class_weight = 'balanced',
        random_state = 42,
        n_jobs       = -1,
    ))
])

rf_pipe.fit(X_train, y_train)
y_pred_rf   = rf_pipe.predict(X_test)
y_proba_rf  = rf_pipe.predict_proba(X_test)[:, 1]

rf_metrics = evaluate_classifier(y_test, y_pred_rf, y_proba_rf,
                                  model_name='Random Forest (default hyperparameters)')
```

5. category_encoded (if included)

Step 5: Interpretation

#Discount and delivery_days remain strong under both measures → they genuinely drive predictive performance.

#Price drops from rank 1 to rank 3 → impurity importance overrated it because continuous variables with many split points often look “important” even if noisy.

#Category_encoded falls out of the top 5 under permutation → showing that while the tree split on it often, shuffling it doesn’t hurt test accuracy much.

👉 Conclusion: Impurity‑based importance can overrate a noisy continuous feature like price because the algorithm finds many possible split thresholds, inflating its apparent importance even if those splits don’t improve generalization.

```python
# Training the pretrained model(resnet18) ,observing for the validation accuracy

import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import models, datasets, transforms
from torch.utils.data import DataLoader, random_split

# 🔹 Preprocessing: replicate grayscale → 3 channels, resize to 224x224, normalize with ImageNet stats
imagenet_mean = [0.485, 0.456, 0.406]
imagenet_std  = [0.229, 0.224, 0.225]

transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.Grayscale(num_output_channels=3),
    transforms.ToTensor(),
    transforms.Normalize(mean=imagenet_mean, std=imagenet_std)
])

# 🔹 Load Fashion-MNIST
train_dataset = datasets.FashionMNIST(root="./data", train=True, download=True, transform=transform)
test_dataset  = datasets.FashionMNIST(root="./data", train=False, download=True, transform=transform)

# Stratified validation split (55k train, 5k val)
train_size, val_size = 55000, 5000
train_dataset, val_dataset = random_split(train_dataset, [train_size, val_size])

batch_size = 64
train_loader = DataLoader(train_dataset, batch_size=batch_size, shuffle=True)
val_loader   = DataLoader(val_dataset, batch_size=batch_size, shuffle=False)
test_loader  = DataLoader(test_dataset, batch_size=batch_size, shuffle=False)

# 🔹 Load pretrained backbone (ResNet-18 example)
backbone = models.resnet18(pretrained=True)

# Freeze all backbone parameters
for param in backbone.parameters():
    param.requires_grad = False

# Replace classifier head with new one for 10 classes
num_features = backbone.fc.in_features
backbone.fc = nn.Linear(num_features, 10)

# 🔹 Optimizer, loss, training config
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(backbone.fc.parameters(), lr=0.001)  # only train head
num_epochs = 5

# 🔹 Feature extraction caching step
def extract_features(loader, model, device):
    model.eval()
    features, labels = [], []
    with torch.no_grad():
        for imgs, lbls in loader:
            imgs = imgs.to(device)
            # Forward pass through frozen backbone up to penultimate layer
            feats = model.avgpool(model.layer4(model.layer3(model.layer2(model.layer1(model.maxpool(model.relu(model.bn1(model.conv1(imgs)))))))))
            feats = torch.flatten(feats, 1)
            features.append(feats.cpu())
            labels.append(lbls)
    return torch.cat(features), torch.cat(labels)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
backbone.to(device)

# Cache features for train/val
train_feats, train_lbls = extract_features(train_loader, backbone, device)
val_feats, val_lbls     = extract_features(val_loader, backbone, device)

# 🔹 Train only the small head on cached features
head = nn.Linear(num_features, 10).to(device)
optimizer = optim.Adam(head.parameters(), lr=0.001)

for epoch in range(num_epochs):
    head.train()
    optimizer.zero_grad()
    outputs = head(train_feats.to(device))
    loss = criterion(outputs, train_lbls.to(device))
    loss.backward()
    optimizer.step()
    print(f"Epoch {epoch+1}/{num_epochs}, Loss: {loss.item():.4f}")

# 🔹 Validation accuracy
head.eval()
with torch.no_grad():
    val_outputs = head(val_feats.to(device))
    _, preds = torch.max(val_outputs, 1)
    acc = (preds == val_lbls.to(device)).float().mean()
    print(f"Validation Accuracy: {acc:.4f}")
```

```python
#validation accuracy is below 50% hence retraing with unfreezing the last layer
#Final Test Accuracy: a scalar (after fine‑tuning, ResNet‑18 typically reaches ~85–90% on Fashion‑MNIST).

# 🔹 Unfreeze the last block (layer4 + fc)
for param in backbone.layer4.parameters():
    param.requires_grad = True
for param in backbone.fc.parameters():
    param.requires_grad = True

# 🔹 Use a lower learning rate for fine-tuning
optimizer = optim.Adam([
    {'params': backbone.layer4.parameters(), 'lr': 1e-4},
    {'params': backbone.fc.parameters(), 'lr': 1e-4}
])

num_epochs_finetune = 5  # longer training for fine-tuning

for epoch in range(num_epochs_finetune):
    backbone.train()
    running_loss, correct, total = 0.0, 0, 0
    for imgs, lbls in train_loader:
        imgs, lbls = imgs.to(device), lbls.to(device)
        optimizer.zero_grad()
        outputs = backbone(imgs)
        loss = criterion(outputs, lbls)
        loss.backward()
        optimizer.step()
        running_loss += loss.item()
        _, preds = torch.max(outputs, 1)
        correct += (preds == lbls).sum().item()
        total += lbls.size(0)
    train_acc = correct / total
    print(f"Epoch {epoch+1}/{num_epochs_finetune}, Loss: {running_loss/len(train_loader):.4f}, Train Acc: {train_acc:.4f}")

    # Validation accuracy
    backbone.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for imgs, lbls in val_loader:
            imgs, lbls = imgs.to(device), lbls.to(device)
            outputs = backbone(imgs)
            _, preds = torch.max(outputs, 1)
            correct += (preds == lbls).sum().item()
            total += lbls.size(0)
    val_acc = correct / total
    print(f"Validation Accuracy: {val_acc:.4f}")
```

```python
import joblib
import numpy as np

def check_return_risk(order_features: dict) -> dict:
    """
    Load the tuned Random Forest model and compute return risk.
    Args:
        order_features (dict): Feature dictionary for one order.
    Returns:
        dict: { 'probability': float, 'bucket': str }
    """

    # 🔹 Load the trained Random Forest model
    model = joblib.load("models/return_risk_model.pkl")

    # Convert dict to feature vector (ensure same order as training)
    X = np.array([list(order_features.values())]).astype(float)

    # 🔹 Predict probability of return (class 1)
    prob = model.predict_proba(X)[0][1]

    # 🔹 Define buckets relative to t*_rf
    # Example: suppose t*_rf was computed in Part 1 as 0.42
    t_rf = 0.42
    low_cut = t_rf
    high_cut = t_rf + 0.15

    if prob < low_cut:
        bucket = "Low"
    elif prob >= high_cut:
        bucket = "High"
    else:
        bucket = "Medium"

    return {"probability": float(prob), "bucket": bucket}
```

👕 T‑shirt/top vs 👔 Shirt
In the confusion matrix, these two categories show a high rate of misclassification. The reason is that both silhouettes are upper‑body garments with similar outlines: short sleeves, a rectangular torso shape, and relatively little texture in grayscale 28×28 images. Fashion‑MNIST images are low‑resolution, so fine details like collars, button plackets, or fabric drape are often lost. As a result, the model sees nearly identical contours and struggles to distinguish whether the item is a casual T‑shirt or a collared shirt. This confusion is plausible because, to the network, both appear as “flat torso coverings with sleeves.”

👢 Ankle boot vs 👟 Sneaker
Another strong confusion pair is between ankle boots and sneakers. Both are footwear silhouettes with a rounded toe box and a vertical shaft rising above the sole. In grayscale, the subtle differences—like the boot’s higher ankle coverage or thicker sole—are blurred. Sneakers with high‑tops look especially similar to ankle boots, and the 28×28 resolution makes it difficult to capture texture cues (laces vs. smooth leather). The model therefore often mistakes one for the other, which is understandable given their overlapping outlines and shared functional design as closed shoes.

These confusion patterns highlight how visual similarity in silhouettes drives misclassification. The categories that differ mainly in fine details (collars, textures, ankle height) are the ones most prone to overlap when downsampled to Fashion‑MNIST’s small grayscale format.

```python
from typing import TypedDict, Optional
from langgraph.graph import StateGraph, END, START

# Define the state schema
class GraphState(TypedDict):
    intent: str
    policy_context: Optional[str]
    tool_output: Optional[str]
    final_answer: Optional[str]

# Dummy node functions
def intent_node(state: GraphState) -> GraphState:
    return {**state, "intent": state.get("intent", "policy")}

def rag_retrieval_node(state: GraphState) -> GraphState:
    return {**state, "policy_context": "Policy KB snippet"}

def tool_calling_node(state: GraphState) -> GraphState:
    return {**state, "tool_output": "Tool result"}

def response_node(state: GraphState) -> GraphState:
    answer = state.get("policy_context") or state.get("tool_output") or "No answer"
    return {**state, "final_answer": f"Response: {answer}"}

# Build the graph
graph = StateGraph(GraphState)

graph.add_node("intent", intent_node)
graph.add_node("rag_retrieval", rag_retrieval_node)
graph.add_node("tool_calling", tool_calling_node)
graph.add_node("response", response_node)

# Entry point: START → intent
graph.add_edge(START, "intent")

# Conditional routing
def route_by_intent(state: GraphState):
    if state["intent"] == "policy":
        return "rag_retrieval"
    elif state["intent"] in ["return_risk", "product_category"]:
        return "tool_calling"
    else:
        return END

graph.add_conditional_edges("intent", route_by_intent)

# Normal edges
graph.add_edge("rag_retrieval", "response")
graph.add_edge("tool_calling", "response")

# Compile
app = graph.compile()

# Test run
result = app.invoke({"intent": "policy"})
print(result["final_answer"])
```
